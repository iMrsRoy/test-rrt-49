```python
#meantemp Guided Lab 330.2.1
import statistics
```

```python
temp=[20, 25, 15]
icecream_sales=[100,120,80]
meanoftemp = statistics.mean(temp)
meanoficecream_sales = statistics.mean(icecream_sales)
print("Mean of temperature is: ", meanoftemp)
print("Mean of ice cream sales is: ", meanoficecream_sales)
```

```python
'''Temperature deviation: X − X̄
Ice cream sales deviation: Y − Ȳ'''

df_TD = [x - meanoftemp for x in temp]
print("Temperature deviation is: ", df_TD)

df_ICD = [y - meanoficecream_sales for y in icecream_sales]
print("Ice cream sales deviation is: ", df_ICD)
```

```python
products = []
for i in range(len(df_TD)):
    products.append(df_TD[i] * df_ICD[i])

print("Products of deviations is: ", products)
```

```python
covariance = sum(products) / (len(temp) - 1)
print("Covariance is: ", covariance)
```
